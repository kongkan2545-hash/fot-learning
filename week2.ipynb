```python
#w2 ct2
#csv
#write
import csv
with open("data.csv", "w", newline="") as f:    #ให้code ไปopen ไฟล์ชื่อว่า data.csv
    writer = csv.writer(f)                      #ให้ตัวแปรรับผลลัพธ์csvที่ใช้คำสั่งwriter โดยเขียนลงในไฟล์ f
    writer.writerow(["name", "age"])            #เพิ่มข้อมูล
    writer.writerow(["Alice", 20])              #เพิ่มข้อมูล

#read    
with open("data.csv", "r") as f:                #ให้ไปอ่านค่าจากไฟล์ที่ชื่อ data.csv
    read = csv.reader(f)                        #กำหนดใช้ฟังก์ชันreaderในการอ่านไฟล์
    for row in read:                            #การใช้ Loop เพื่อไล่อ่านข้อมูลในไฟล์ทีละแถว
        print(row)                              #นำค่าที่อ่านได้มาแสดง
```

```python
#w2 ct3
#dict/json
import json
sensor = {
"device_id" : "box01",
"temp" : 29.5,
"humi" : 71
}

print(sensor)                                   #แสดงค่าทั้งหมด
print(sensor["temp"])                           #เลือกแสดงแค่ค่าtemp

#แปลง Python → JSON
json_data = json.dumps(sensor)                  #แปลง sensorให้กลายเป็น String (ข้อความ) ในรูปแบบ JSON
print(type(json_data))                          #แสดงผลเป็น String
           
#แปลง JSON → Python
python_data = json.loads(json_data)             #แปลง sensorที่เป็น JSON กลับมาเป็น Python
print(python_data)                              #แสดงผล
```
